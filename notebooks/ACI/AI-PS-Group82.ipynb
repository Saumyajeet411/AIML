3.	Use appropriate data structures and implement given informed and local search algorithm and Print the path taken by the agent. [3 +3 = 6 Marks]

```python
# -------------------------------
# Flood Relief Assignment
# Greedy Best First Search + Genetic Algorithm
# -------------------------------

import heapq, time, tracemalloc
import random

# Graph definition
graph = {
    "Nungambakkam": {"Guindy": 12, "Perumbakkam": 10, "Tambaram": 15},
    "Guindy": {"Velachery": 14, "Perumbakkam": 7},
    "Velachery": {"Perumbakkam": 6},
    "Perumbakkam": {"Tambaram": 11, "Purasawalkam": 8},
    "Tambaram": {"Purasawalkam": 9},
    "Purasawalkam": {}
}

# -------------------------------
# Greedy Best First Search
# -------------------------------
def greedy_best_first(start):
    pq = []
    heapq.heappush(pq, (0, start, [start]))
    visited_edges = set()
    path = []

    while pq:
        cost, node, path_so_far = heapq.heappop(pq)
        path = path_so_far
        for neighbor, edge_cost in graph[node].items():
            edge = tuple(sorted([node, neighbor]))
            if edge not in visited_edges:
                visited_edges.add(edge)
                heapq.heappush(pq, (edge_cost, neighbor, path + [neighbor]))
    return path

# -------------------------------
# Genetic Algorithm
# -------------------------------
vertices = list(graph.keys())
edges = {(u,v):w for u in graph for v,w in graph[u].items()}

def fitness(path):
    cost, visited = 0, set()
    for i in range(len(path)-1):
        u, v = path[i], path[i+1]
        if (u,v) in edges:
            cost += edges[(u,v)]
            visited.add(tuple(sorted([u,v])))
        else:
            cost += 100  # penalty for invalid edge
    return 1 / (cost + (len(edges)-len(visited))*50)

def mutate(path):
    i, j = random.sample(range(len(path)), 2)
    path[i], path[j] = path[j], path[i]
    return path

def crossover(p1, p2):
    cut = random.randint(1, len(p1)-2)
    child = p1[:cut] + [v for v in p2 if v not in p1[:cut]]
    return child

def genetic_algorithm(start, generations=50, pop_size=10):
    population = [[start] + random.sample(vertices, len(vertices)) for _ in range(pop_size)]
    for _ in range(generations):
        scores = [(fitness(p), p) for p in population]
        scores.sort(reverse=True)
        population = [p for _, p in scores[:pop_size//2]]
        while len(population) < pop_size:
            p1, p2 = random.sample(population, 2)
            child = crossover(p1, p2)
            if random.random() < 0.3:
                child = mutate(child)
            population.append(child)
    best = max(population, key=fitness)
    return best

# -------------------------------
# Run both algorithms
# -------------------------------
if __name__ == "__main__":
    print("Greedy Best First Search Path:")
    print(greedy_best_first("Nungambakkam"))

    print("\nGenetic Algorithm Path:")
    print(genetic_algorithm("Nungambakkam"))
```

4.	Find and print space and time complexity using code in your implementation. [3 Marks ]

```python

# Measure time and space for GBFS
tracemalloc.start()
start_time = time.time()
gbfs_path = greedy_best_first("Nungambakkam")
end_time = time.time()
current, peak = tracemalloc.get_traced_memory()
tracemalloc.stop()

print("Greedy Best First Search Path:", gbfs_path)
print("Time Complexity (measured):", end_time - start_time, "seconds")
print("Space Complexity (measured):", peak, "bytes")




# Measure time and space for Genetic Algorithm
tracemalloc.start()
start_time = time.time()
ga_path = genetic_algorithm("Nungambakkam")
end_time = time.time()
current, peak = tracemalloc.get_traced_memory()
tracemalloc.stop()

print("\nGenetic Algorithm Path:", ga_path)
print("Time Complexity (measured):", end_time - start_time, "seconds")
print("Space Complexity (measured):", peak, "bytes")
```
